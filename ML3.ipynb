```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# standard BuyComputer dataset (14 instances)
data = [
    {"Age": "<=30",   "Income": "High",   "Student": "No",  "Credit": "Fair",      "BuyComputer": "No"},
    {"Age": "<=30",   "Income": "High",   "Student": "No",  "Credit": "Excellent", "BuyComputer": "No"},
    {"Age": "31..40", "Income": "High",   "Student": "No",  "Credit": "Fair",      "BuyComputer": "Yes"},
    {"Age": ">40",    "Income": "Medium", "Student": "No",  "Credit": "Fair",      "BuyComputer": "Yes"},
    {"Age": ">40",    "Income": "Low",    "Student": "Yes", "Credit": "Fair",      "BuyComputer": "Yes"},
    {"Age": ">40",    "Income": "Low",    "Student": "Yes", "Credit": "Excellent", "BuyComputer": "No"},
    {"Age": "31..40", "Income": "Low",    "Student": "Yes", "Credit": "Excellent", "BuyComputer": "Yes"},
    {"Age": "<=30",   "Income": "Medium", "Student": "No",  "Credit": "Fair",      "BuyComputer": "No"},
    {"Age": "<=30",   "Income": "Low",    "Student": "Yes", "Credit": "Fair",      "BuyComputer": "Yes"},
    {"Age": ">40",    "Income": "Medium", "Student": "Yes", "Credit": "Fair",      "BuyComputer": "Yes"},
    {"Age": "<=30",   "Income": "Medium", "Student": "Yes", "Credit": "Excellent", "BuyComputer": "Yes"},
    {"Age": "31..40", "Income": "Medium", "Student": "No",  "Credit": "Excellent", "BuyComputer": "Yes"},
    {"Age": "31..40", "Income": "High",   "Student": "Yes", "Credit": "Fair",      "BuyComputer": "Yes"},
    {"Age": ">40",    "Income": "Medium", "Student": "No",  "Credit": "Excellent", "BuyComputer": "No"},
]

df = pd.DataFrame(data)
features = ["Age", "Income", "Student", "Credit"]
target_col = "BuyComputer"
positive_val = "Yes"

print("Dataset size:", df.shape)
display(df)
print("
Target counts:
", df[target_col].value_counts())
```

```python
# find-s: start with most specific hypothesis, generalize only on positive examples
pos_df = df[df[target_col] == positive_val].reset_index(drop=True)
print(f"Total positive examples to process: {len(pos_df)}")

# specific hypothesis initialization
h = ["Phi"] * len(features)
history = []

for idx, row in pos_df.iterrows():
    example = [row[f] for f in features]
    
    if h == ["Phi"] * len(features):
        # initialize with first positive example
        h = list(example)
    else:
        # generalize whenever values differ
        for i in range(len(features)):
            if h[i] != example[i]:
                h[i] = "?"
                
    history.append(list(h))

# tabulate hypothesis evolution
hist_df = pd.DataFrame(history, columns=features)
hist_df.index = [f"Positive {i+1}" for i in range(len(pos_df))]

print("
Hypothesis Evolution across positive examples:")
display(hist_df)

print("
Final Hypothesis:")
for f, v in zip(features, h):
    print(f"  {f} = {v}")
```

```python
# visualize how specific features turn to '?' (generalize) over steps
# encode: specific value = 1, generalized '?' = 0
encoded_hist = hist_df.map(lambda x: 0 if x == "?" else 1)

plt.figure(figsize=(8, 4))
sns.heatmap(encoded_hist, annot=hist_df.values, fmt="", cmap="Blues_r", cbar=False, linewidths=1)
plt.title("Find-S Hypothesis Generalization (Blue = Specific, White = Generalized ?)")
plt.xlabel("Features")
plt.ylabel("Step")
plt.tight_layout()
plt.show()
```
