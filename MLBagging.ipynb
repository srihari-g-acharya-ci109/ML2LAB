```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def generate_samples_until_convergence(n_elements=10, val_range=(1, 10), window_size=50, threshold=0.001, max_samples=1000):
    samples = {}
    unique_percentages = []
    running_averages = []

    i = 1
    converged = False

    while not converged and i <= max_samples:
        # draw bootstrap sample with replacement
        sample = np.random.choice(range(val_range[0], val_range[1] + 1), size=n_elements, replace=True)
        sample_name = f"D{i}"
        samples[sample_name] = sample

        # unique element percentage
        unique_pct = (len(np.unique(sample)) / n_elements) * 100
        unique_percentages.append(unique_pct)

        # rolling average over window
        if i >= window_size:
            current_avg = np.mean(unique_percentages[-window_size:])
            running_averages.append(current_avg)

            if len(running_averages) > 2:
                diff = abs(running_averages[-1] - running_averages[-2])
                if diff < threshold:
                    converged = True
                    print(f"Convergence reached at sample {sample_name} (change: {diff:.5f}%)")
        else:
            running_averages.append(np.mean(unique_percentages))

        i += 1

    return samples, unique_percentages, running_averages

# simulate convergence to theoretical ~63.2%
samples, unique_pcts, running_avgs = generate_samples_until_convergence()

print(f"Total samples drawn: {len(samples)}")
print(f"Final running average of unique items: {running_avgs[-1]:.2f}%")
```

```python
# plot individual sample percentages and running average
plt.figure(figsize=(13, 5))
indices = list(range(1, len(unique_pcts) + 1))

plt.scatter(indices, unique_pcts, color='skyblue', alpha=0.5, label='Unique % per sample')
plt.plot(indices, running_avgs, color='crimson', linewidth=2, label='Running Average')
plt.axhline(63.2, color='black', linestyle='--', alpha=0.7, label='Theoretical 63.2%')

plt.title('Bootstrap Sampling: Percentage of Unique Elements over Iterations')
plt.xlabel('Sample index (D1, D2, ...)')
plt.ylabel('Unique Elements (%)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
